## Learning Objectives

- Setup Opencode with Nemotron 3.5 Lightning as the coding assistant

Configure the local endpoint set up in [01_inference_endpoint](./01_inference_endpoint.ipynb)

````python
%%bash

export INF_URL="http://127.0.0.1:8000/v1"
echo $INF_URL

cat > ~/opencode.json <<EOF
{
  "\$schema": "https://opencode.ai/config.json",
  "provider": {
    "lm-studio": {
      "npm": "@ai-sdk/openai-compatible",
      "name": "Local NIM",
      "options": {
        "baseURL": "${INF_URL}"
      },
      "models": {
        "nvidia/nemotron-3.5-lightning": {
          "name": "nemo-3.5-lightning"
        }
      }
    }
  }
}
EOF
````

Open a new terminal in jupyter lab by clicking on File -> New -> Terminal

<div style="text-align: center;">
  <img src="images/jupyterlab-new-terminal.png" style="width: 900px; height: auto;">
</div>

In the terminal window, first install OpenCode if not done already using

```bash
curl -fsSL https://opencode.ai/v2/install | bash
source ~/.bashrc
```
Then input `opencode` and press enter. This will get you into the opencode session. If `opencode` is not installed, install it using `curl -fsSL https://opencode.ai/install | bash`.

<div style="text-align: center;">
  <img src="images/opencode.png" style="width: 900px; height: auto;">
</div>

Input `/models` and select `nemo-3.5-lightning`. You can also choose any other model available in the catalog.

Opencode is now setup. You can use this to ask any question on the tutorials or assist you in the challenges.

Example prompt: ``Explain the code in ../tutorial/solutions/mcp/mcp_server_http.py``
