# Agentic AI for Research Workflows

Coding agents can now carry out real research work: fetching data, cleaning it, fitting models, and building interactive outputs from plain-language instructions. 

That power comes with a catch: every detail you don't specify, the agent decides for you.

In the next notebooks, we build one project end to end: predicting which flights out of SFO, OAK, and SJC will arrive late, using real government data. Along the way we practice the habits that make agent-assisted research trustworthy: inspecting the agent's steps, pinning down specifications, checking results against baselines, and turning a pile of conversations into a reproducible pipeline.

## One Sentence Goal

Download last month's flight delay data from the Bureau of Transportation Statistics, build a model that predicts which flights run late, and make a page that shows what you found.

## Create Your Project

We're starting from scratch — a fresh project that will hold everything we build today.

1. Open a terminal.
2. Create a new folder called `flight-delays` and navigate to it: `mkdir flight_delays`, `cd flight-delays`
3. Open `opencode`

That's it. The agent's world (or workspace) is now that folder.

💡 **Tip:** Not sure where the folder ended up on your computer? Ask the agent: "What folder are we in? Give me the full path."

## Conversations

Inside a project, your work is organized into **conversations** or **sessions**. 

The paradigm to internalize is **one conversation = one task** that you want the agent to do.

Downloading the data is a task - that's a conversation. Exploring the data is a task - that's a conversation. Fitting a model - conversation. When you move on to a new task, don't keep piling into one endless conversation. Just start a new conversation.

There are two things to keep in mind:

- **Files persist across conversations.** Anything the agent created in an earlier conversation is still in the folder, and a new conversation can see it.
- **The chat history doesn't.** A new conversation starts fresh. The agent doesn't remember what you discussed in the last one. However, the agent *can* access the project's files, instructions, shared sources, and memories from earlier chats (if enabled). In other words, if information lives in your head, the agent doesn't have access to it unless you tell it. But if it exists in a file, the agent can search for it and use it.

🔔 **Question:** If each conversation starts with no memory of the previous ones, how does the agent know the rules of your project? 

## Permissions

Before we set the agent loose, it's worth knowing what it's allowed to do. How much freedom the agent has is a setting.

You can access the `opencode` Permissions setting using the `/settings` commands.

- **`Prompt`** - always ask before editing external files or using the internet.
- **`Auto accept`** - the agent reviews its own requests, and only asks about actions it detects as potentially unsafe. Fewer pauses, but its self-review can make mistakes.

⚠️ **Warning:** Set it to `Prompt` today, and check what your current setting is — it may not be the cautious one. `Auto accept` means the agent can touch and do anything on your computer with no review, which sometimes can lead to unintended consequences.

You will also notice OpenCode include 2 built-in agents you can switch between with the `maj/shift + tab` keys.

- **`build`** - Default, full-access agent for development work.
- **`plan`** - Read-only agent for anaysis and code exploration. Denies file edits by default, asks permission before running bash commands, ideal for exploring an unfamiliar code base.

# Give Your Agent Rules

Agents follow instructions you put in a special file called `AGENTS.md`. Think of it as a note pinned to the wall of the project: the agent reads it every time it starts working.

Copy and paste this prompt into OpenCode:

```
Create a file called AGENTS.md in this project with exactly this content:

# Project rules

## Environment
- Install packages with pip or conda into a dedicated virtual environment in this folder.

## Data
- Data files live in the data/ folder.
- Never modify raw downloaded files. Save cleaned versions as new files.
- After any filtering step, state how many rows remain.

## Results
- Store results in the results/ folder.

## Analysis
- Use random_state=0 anywhere randomness is involved.
- When reporting a model's accuracy, always also report the accuracy of predicting the most common outcome for every observation.

## Workflow
- Propose a plan before creating or editing files.
- After running a command, show the exact command and summarize the result.
```

💡 **Tip:** `AGENTS.md` can steer model behavior, but it is not deterministic. Instructions can get lost, especially in long-running conversations where the context gets bloated.

# Key Points

- A **coding agent** can take actions on your computer, like reading files, writing code, and running code.
- The agent works inside a project folder, and everything it does is inspectable.
- Keep **permissions** on ask-before-acting while you're learning.
- `AGENTS.md` holds project rules your agent follows in every conversation.
- If you don't know what to ask, ask the agent what to ask.
